Last week the computer did three things: evaluate, iterate, find a fixed point. This week we find the fixed point *directly*, as a root of $r(k) = g(k) - k$, and then use the same idea to solve our first general-equilibrium model.

The running example is the Solow residual from last lecture, because we know the answer:
$$r(k) = s\,k^{\alpha} - (\delta + n + g)\,k, \qquad r(k^*) = 0,\quad k^* \approx 1.589.$$

```python
import numpy as np
import matplotlib.pyplot as plt

def solow_residual(k, alpha=0.33, s=0.30, n=0.03, g=0.04, delta=0.15):
    return s * np.power(k, alpha) - (delta + n + g) * k

def solow_residual_prime(k, alpha=0.33, s=0.30, n=0.03, g=0.04, delta=0.15):
    return alpha * s * np.power(k, alpha - 1) - (delta + n + g)

kstar = (0.30 / 0.22) ** (1 / 0.67)
k = np.linspace(0.01, 4, 400)
plt.plot(k, solow_residual(k)); plt.axhline(0, color="grey"); plt.xlabel("k"); plt.ylabel("r(k)"); plt.show()
```

# Lecture 2: root finding, and a first equilibrium

**ECU44291 Quantitative Macroeconomics, Week 2.**

Last week the computer did three things: evaluate, iterate, find a fixed point. This week we find the fixed point *directly*, as a root of $r(k) = g(k) - k$, and then use the same idea to solve our first general-equilibrium model.

The running example is the Solow residual from last lecture, because we know the answer:
$$r(k) = s\,k^{\alpha} - (\delta + n + g)\,k, \qquad r(k^*) = 0,\quad k^* \approx 1.589.$$

```python
import numpy as np
import matplotlib.pyplot as plt

def solow_residual(k, alpha=0.33, s=0.30, n=0.03, g=0.04, delta=0.15):
    return s * np.power(k, alpha) - (delta + n + g) * k

def solow_residual_prime(k, alpha=0.33, s=0.30, n=0.03, g=0.04, delta=0.15):
    return alpha * s * np.power(k, alpha - 1) - (delta + n + g)

kstar = (0.30 / 0.22) ** (1 / 0.67)
k = np.linspace(0.01, 4, 400)
plt.plot(k, solow_residual(k)); plt.axhline(0, color="grey"); plt.xlabel("k"); plt.ylabel("r(k)"); plt.show()
```

## Task 1: bisection

Write `bisection(f, a, b, tol=1e-8, max_iter=200)` returning `(root, brackets)`, where `brackets` is the list of `(a, b)` intervals visited. It must return the endpoint if `f` is already zero there, and raise `ValueError` if `f(a)` and `f(b)` have the same sign. Find $k^*$ on $[0.5, 4]$ and count the iterations. Why 29? Compare with $\log_2(3.5/10^{-8})$. Then try the bracket $[0, 4]$: which root do you get, and why?

```python
# SOLUTION
def bisection(f, a, b, tol=1e-8, max_iter=200):
    fa, fb = f(a), f(b)
    if fa == 0:                             # already at a root
        return a, [(a, b)]
    if fb == 0:
        return b, [(a, b)]
    if fa * fb > 0:
        raise ValueError("no bracket: f(a) and f(b) have the same sign")
    brackets = []
    for _ in range(max_iter):               # loop
        brackets.append((a, b))
        m = 0.5 * (a + b)                   # update
        fm = f(m)
        if fm == 0 or 0.5 * (b - a) < tol:  # stopping rule
            return m, brackets
        if fa * fm < 0:
            b, fb = m, fm
        else:
            a, fa = m, fm
    raise RuntimeError("no convergence")

root, br = bisection(solow_residual, 0.5, 4.0)
print(f"root {root:.8f}, k* {kstar:.8f}, {len(br)} iterations; log2(3.5 / 1e-8) = {np.log2(3.5 / 1e-8):.1f}")
print(f"bracket [0, 4]: {bisection(solow_residual, 0.0, 4.0)[0]}")
```

## Task 2: Newton, and how it fails

Write `newton(f, x0, fprime, tol=1e-10, max_iter=50)` returning `(root, iterates)`. It must raise `RuntimeError` if an iterate is not finite. Run it from $x_0 = 3.5$ and count iterations. Then run it from $x_0 = 0.1$. Read the error. Draw the tangent at $0.1$ on paper and say what happened.

```python
# SOLUTION
def newton(f, x0, fprime, tol=1e-10, max_iter=50):
    x, iterates = x0, [x0]
    for _ in range(max_iter):
        fx, d = f(x), fprime(x)
        if not (np.isfinite(fx) and np.isfinite(d)):
            raise RuntimeError(f"newton left the domain at x = {float(x):.4g}")
        x_new = x - fx / d
        iterates.append(x_new)
        if abs(x_new - x) < tol:
            return x_new, iterates
        x = x_new
    raise RuntimeError("no convergence")

root, its = newton(solow_residual, 3.5, solow_residual_prime)
print(f"from 3.5: root {root:.10f} in {len(its) - 1} steps")
try:
    newton(solow_residual, 0.1, solow_residual_prime)
except RuntimeError as e:
    print("from 0.1:", e)
```

## Task 3: secant, then the library

Write `secant(f, x0, x1, ...)`: Newton with the derivative replaced by the slope through the last two points. Then call `scipy.optimize.brentq(solow_residual, 0.5, 4.0)`. Brent's method is bisection's safety with the secant's speed; from now on it is what you call when you have a bracket.

```python
# SOLUTION (homework: held back from the posted notebook until next week)
def secant(f, x0, x1, tol=1e-10, max_iter=50):
    f0, f1 = f(x0), f(x1)
    its = [x0, x1]
    for _ in range(max_iter):
        x2 = x1 - f1 * (x1 - x0) / (f1 - f0)
        its.append(x2)
        if abs(x2 - x1) < tol:
            return x2, its
        x0, f0, x1, f1 = x1, f1, x2, f(x2)
    raise RuntimeError("no convergence")

root, its = secant(solow_residual, 3.0, 3.5)
print(f"secant: {root:.10f} in {len(its) - 2} steps")
```

```python
# SOLUTION
from scipy.optimize import brentq

print(f"brentq: {brentq(solow_residual, 0.5, 4.0):.10f}")
```

## Task 4: the two-period OLG economy

Young households earn $w$, pay labour tax $\tau_L$ and a pension contribution $\tau_p$, save $s$; old households consume $R s$ plus the pension. Firms, with $L = N_y$: $q = \alpha A K^{\alpha-1} L^{1-\alpha}$, $w = (1-\alpha) A K^{\alpha} L^{-\alpha}$, and $R = 1 + (1-\tau_K)(q - \delta)$. With log utility the young save
$$s(w, R) = \frac{\beta R\,[(1-\tau_L-\tau_p)w] - \text{pension}}{(1+\beta) R}, \qquad \text{pension} = \tau_p w \frac{N_y}{N_o}.$$
Equilibrium: $K = N_y\, s(w(K), R(K))$. Write `prices(K, p)`, `savings(w, R, p)`, `residual(K, p)`, plot the residual on $[0, 0.6]$, and find $K^*$ with `brentq`. Why must the bracket exclude $K = 0$?

```python
# SOLUTION
p = dict(alpha=0.3, A=1.0, beta=0.9, delta=0.0, tau_L=0.2,
         tau_K=0.15, tau_p=0.0, N_y=1.0, N_o=1.0)

def prices(K, p):
    L = p["N_y"]                                                           # labour = the young
    q = p["alpha"] * p["A"] * K ** (p["alpha"] - 1) * L ** (1 - p["alpha"])   # rental rate
    w = (1 - p["alpha"]) * p["A"] * K ** p["alpha"] * L ** (-p["alpha"])      # wage
    R = 1 + (1 - p["tau_K"]) * (q - p["delta"])                              # household's gross return
    return q, w, R

print(prices(0.15, p))
```

```python
# SOLUTION
def savings(w, R, p):
    pension = p["tau_p"] * w * p["N_y"] / p["N_o"]
    return (p["beta"] * R * (1 - p["tau_L"] - p["tau_p"]) * w
            - pension) / ((1 + p["beta"]) * R)

q, w, R = prices(0.15, p)
print(savings(w, R, p))
```

```python
# SOLUTION
def residual(K, p):
    if K <= 0:
        return 0.0                          # no capital, no wage, nobody saves: the trivial equilibrium
    _, w, R = prices(K, p)
    return p["N_y"] * savings(w, R, p) - K

print(residual(0.05, p), residual(0.5, p))                  # opposite signs: a bracket

KK = np.linspace(0.001, 0.6, 300)
plt.plot(KK, [residual(K, p) for K in KK]); plt.axhline(0, color="grey"); plt.xlabel("K"); plt.show()
Kstar = brentq(residual, 1e-3, 2.0, args=(p,))
print(f"K* = {Kstar:.5f};  residual near K = 0 is {residual(1e-12, p):.1e} and goes to zero: K = 0 is the trivial root")
print(f"brentq on [0, 2]: {brentq(residual, 0.0, 2.0, args=(p,))}")
```

## Task 5: a pay-as-you-go pension

Raise $\tau_p$ from 0 to 0.3 and plot $K^*$. Explain the direction in one sentence about what the young are being promised. This is the first policy experiment of the module; the project's is the same idea in a model with sixty cohorts.

```python
# SOLUTION
taus = np.linspace(0, 0.3, 16)
Ks = [brentq(residual, 1e-3, 2.0, args=({**p, "tau_p": t},)) for t in taus]
plt.plot(taus, Ks, "-o"); plt.xlabel("contribution rate"); plt.ylabel("K*"); plt.show()
print(f"K* falls from {Ks[0]:.4f} to {Ks[-1]:.4f}")
```
